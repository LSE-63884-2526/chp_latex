# CASE CHECK

```python
import os
import numpy as np
import pandas as pd
from sklearn.metrics import normalized_mutual_info_score

EMB_DIR = r"C:\Users\user\Desktop\Thesis\Embeddings"
TERM, STEM, WIN, TAG = "kadın", "kadın", 15, "_maskwhite"

meta = pd.read_csv(os.path.join(EMB_DIR, f"{TERM}_cw{WIN}metadata{TAG}.csv"))

def tr_lower(t):
    return (str(t).replace("İ", "i").replace("I", "ı").replace("Ğ", "ğ")
            .replace("Ş", "ş").replace("Ç", "ç").replace("Ö", "ö").replace("Ü", "ü").lower())

def case_of(s, stem=STEM):
    s = s.split(" | ")[0].strip("'’")
    suf = s[len(stem):] if s.startswith(stem) else s
    if suf == "":                                              return "nom"
    if suf.endswith(("dan", "den", "tan", "ten")):             return "abl"
    if suf.endswith(("nin", "nın", "nun", "nün",
                     "in", "ın", "un", "ün")):                 return "gen"
    if suf.endswith(("da", "de", "ta", "te")):                 return "loc"
    if suf.endswith(("la", "le")):                             return "ins"
    if suf.endswith(("a", "e")):                               return "dat"
    if suf.endswith(("i", "ı", "u", "ü")):                     return "acc"
    if suf.endswith(("imiz", "ımız")):                         return "nom"
    if suf.endswith(("dir", "dır", "tir", "tır", "iz", "ız")): return "pred"
    return "nom"

def case_check(meta, labels):
    m = meta.copy()
    assert len(m) == len(labels), f"meta {len(m)} vs labels {len(labels)}"
    m["cluster"] = np.asarray(labels)
    m["case"] = m["surface"].apply(tr_lower).apply(case_of)
    m = m[m["cluster"] != -1]                      # drop HDBSCAN noise

    nmi = normalized_mutual_info_score(m["case"], m["cluster"])
    tab = pd.crosstab(m["cluster"], m["case"], normalize="index")
    print(f"NMI(case, cluster): {nmi:.3f}")
    print(tab.max(axis=1).describe().round(2))    # cluster purity
    return tab.round(2), m
```
